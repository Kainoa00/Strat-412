## Part 1: Sprouts Farmers Market — 10-K Research
*(NASDAQ: SFM — Annual Reports available at investors.sprouts.com)*

### Item 1: Business Overview

| Metric | Detail |
|--------|--------|
| Founded | 2002 (Chandler, AZ) |
| IPO | August 2013 (NASDAQ: SFM) |
| Store count (2024) | ~487 stores |
| States | ~24 states |
| Store size | 21,000–25,000 sq ft (smaller than traditional grocery) |
| Annual new openings | ~35–45 stores/year (~10% unit growth rate) |
| Private label revenue | ~25% of total revenue |
| Target customer | "Health enthusiasts" — value-conscious shoppers prioritizing fresh, natural, organic |

### Key Strategic Pillars (from 10-K Item 1)
1. **Differentiated assortment**: ~18,000–20,000 SKUs, emphasis on fresh produce ("farmer's market feel"), vitamins/supplements, bulk foods
2. **Value positioning**: Positioned *below* Whole Foods on price — organic at accessible prices
3. **Small-format, suburban focus**: Targets suburban strip centers and lifestyle centers (not urban cores)
4. **Geographic expansion**: Heavy TX, FL, AZ presence; expanding East (PA, MD, VA, NC, SC, DE)
5. **Private label growth**: Sprouts brand products growing faster than national brands

### Item 1A: Key Risk Factors (from 10-K)
- **Competition**: WFM/Amazon, Trader Joe's, Kroger Simple Truth, Target Good & Gather, Aldi
- **Perishables risk**: ~50% of sales are perishables (produce, meat, seafood) → high shrink risk
- **Supplier concentration**: Relies on regional and specialty suppliers who may be disrupted
- **Real estate**: Dependent on identifying suitable suburban locations as chain expands East
- **Consumer perception**: Must maintain "value" positioning as it moves into higher-cost markets

### Growth Trajectory (store count by year — approximate)

| 2013 (IPO) | ~157 |
| 2016 | ~240 |
| 2019 | ~340 |
| 2022 | ~390 |
| 2024 | ~487 |

**Growth pattern**: Organic (no major acquisitions). All stores are new builds or tenant improvements.

## Part 2: Trader Joe's — Private Company Research
*(Private — no SEC filings. Sources: Freakonomics podcast, press coverage)*

### Freakonomics Podcast: "What's the Secret of Trader Joe's?"
*Episode reference: Freakonomics Radio, originally aired 2015, updated 2022*

#### Founding & Origin Story
- **Founded**: 1967 by **Joe Coulombe** in Pasadena, CA (converted from Pronto Markets convenience stores)
- **Original insight**: Jet travel was making Americans more worldly/adventurous with food tastes
- **Target demographic (Coulombe's phrase)**: *"Overeducated and underpaid"* — teachers, professors, grad students, artists who are intellectually sophisticated but budget-constrained
- **Aldi acquisition**: Sold to German discount chain **Aldi Nord** in **1979** (Theo Albrecht family)
  - Aldi connection gives massive private-label buying power
  - Trader Joe's operates independently under the Aldi Nord umbrella (Aldi Sud owns US Aldi stores separately)

#### Core Strategy
| Element | Detail |
|---------|--------|
| SKUs | ~3,000–4,000 (vs. 30,000+ at traditional grocery) |
| Private label | ~80% of products are Trader Joe's brand |
| Advertising | **None** — zero traditional advertising budget |
| Loyalty program | **None** |
| Store size | ~10,000–15,000 sq ft (very small format) |
| Pricing | Low — competitive with Aldi, below Whole Foods/Sprouts |
| Locations | Dense urban + college towns + affluent suburbs |

#### Why the Low SKU Count Works
- Higher volume per SKU → better supplier negotiation leverage
- Less shelf complexity → lower labor/restocking costs
- Creates **scarcity psychology** — customers buy what's there; seasonal items create excitement
- Each product must be approved by the CEO (historically) — very curated

#### Geographic Expansion History
| Period | Expansion |
|--------|----------|
| 1967–1979 | California only |
| 1979–1995 | CA + AZ + Pacific Northwest |
| 1996 | First East Coast store (Brookline, MA) — targeting Harvard/MIT corridor |
| 2000s | Rapid East Coast expansion (NYC, Boston, DC) |
| 2010s | National coverage; ~42 states by 2020 |
| 2024 | ~637 stores, all 50 states |

**Growth pattern**: Very deliberate — cluster stores in walkable urban areas and college towns. Rarely in strip malls. Often in converted buildings.

#### Competitive Moat (from podcast analysis)
1. **Cult brand loyalty**: "Trader Joe's fans" are evangelical — free word-of-mouth marketing
2. **Private label at scale**: 80% private label → controls pricing and quality without brand premium
3. **Employee culture**: Above-market pay for retail (starting ~$18-20/hr), crew members wear Hawaiian shirts, managers called "Captains" and "First Mates"
4. **No online ordering**: Forces in-store discovery — intentional friction to drive basket size

## Part 3: Whole Foods Market — 10-K Research
*(Amazon subsidiary since 2017; previously NASDAQ: WFM)*

### Business Overview

| Metric | Detail |
|--------|--------|
| Founded | 1980 (Austin, TX) — merger of SaferWay and Clarksville Natural Grocery |
| Founders | John Mackey, Renée Lawson Hardy, Craig Weller, Mark Skiles |
| Peak as public company | 2017 — ~470 stores, $16B revenue |
| Amazon acquisition | **June 2017** — **$13.7 billion** (bought entire company including Prime Now integration) |
| Store count (2024) | ~530–550 stores |
| Store size | 30,000–50,000 sq ft (large format) |
| Target customer | Affluent, health-conscious urban/suburban shoppers |
| Nickname | "Whole Paycheck" — perceived as premium/expensive |

### Growth History: Acquisition-Driven Expansion
Whole Foods grew primarily through **acquiring regional natural food chains** — very different from Sprouts (organic growth) and Trader Joe's (deliberate new store opens).

| Year | Acquisition | Stores Added |
|------|-------------|-------------|
| 1992 | Bread of Life (Boston) | 2 |
| 1996 | Fresh Fields (East Coast) | 22 |
| 1996 | Bread & Circus (New England) | 7 |
| 1997 | Merchant of Vino (Michigan) | 6 |
| 1999 | Natural Abilities / Amrion | — |
| 2004 | Fresh & Wild (UK) | 7 |
| 2007 | Wild Oats Markets | 70+ |
| 2017 | Acquired BY Amazon ($13.7B) | — |

**Post-Amazon changes**:
- Amazon Prime discounts in-store
- Amazon Lockers in stores
- Prime Now same-day delivery from WFM locations
- Slower new store growth; closed some underperforming locations
- 2023-2024: Renewed expansion push (30+ new stores planned)

### Strategic Positioning Post-Amazon
1. **Amazon flywheel integration**: WFM stores serve as last-mile fulfillment nodes for Amazon Fresh/Prime
2. **Premium private label**: 365 by Whole Foods brand — their lower-price tier (started 2016 as standalone stores, converted to in-store brand)
3. **Quality standards**: 100+ banned ingredients/additives — strict product quality standards
4. **Urban flagship focus**: Large-format flagship stores (45,000+ sq ft) in major metros act as brand showrooms

### Risk Factors (from pre-2017 10-K, still relevant)
- **Price perception**: "Whole Paycheck" reputation limits mass-market appeal
- **Competition from conventional grocery**: Kroger, Target, Costco all expanded organic/natural sections
- **Amazon cannibalization**: Amazon Fresh competes with WFM delivery in same markets
- **Real estate costs**: Large-format stores require prime (expensive) urban real estate

```python
# ─────────────────────────────────────────────
# CODE: Build Historical Store Count Chart
# ─────────────────────────────────────────────
# Data sourced from 10-K filings, press releases, and company history.
# Use this to create Tableau Slide 2 — Growth Trajectories.

import pandas as pd
import matplotlib.pyplot as plt

# Approximate historical store counts (from public filings and press)
history = {
    "Year": [2002, 2005, 2008, 2010, 2012, 2013, 2015, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    "Sprouts":     [1,    5,   20,   40,   90,  157,  204,  280,  313,  340,  362,  374,  382,  407,  487],
    "Trader Joes": [140, 200, 290,  350,  400,  430,  474,  495,  503,  510,  530,  560,  590,  610,  637],
    "Whole Foods": [130, 175, 270,  300,  335,  362,  431,  479,  493,  500,  510,  511,  527,  537,  550],
}

df_history = pd.DataFrame(history)

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df_history["Year"], df_history["Sprouts"],     marker="o", label="Sprouts",      color="#4CAF50", linewidth=2)
ax.plot(df_history["Year"], df_history["Trader Joes"], marker="s", label="Trader Joe's", color="#E53935", linewidth=2)
ax.plot(df_history["Year"], df_history["Whole Foods"], marker="^", label="Whole Foods",  color="#1E88E5", linewidth=2)

# Annotate Amazon acquisition
ax.axvline(x=2017, color="gray", linestyle="--", alpha=0.6)
ax.text(2017.1, 520, "Amazon acquires\nWhole Foods ($13.7B)", fontsize=9, color="gray")

ax.set_title("Store Count Growth: Sprouts vs Trader Joe's vs Whole Foods", fontsize=14, fontweight="bold")
ax.set_xlabel("Year")
ax.set_ylabel("Number of Stores")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("growth_trajectory.png", dpi=150, bbox_inches="tight")
plt.show()
print("Chart saved as growth_trajectory.png")

# Also export as CSV for Tableau
df_history_melted = df_history.melt(id_vars="Year", var_name="Chain", value_name="Store_Count")
df_history_melted.to_csv("store_count_history.csv", index=False)
print("Data exported to store_count_history.csv")
```
